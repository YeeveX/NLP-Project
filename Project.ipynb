````python
class MessageHistory:
    def __init__(self):
        self.messages = []

    def add_message(self, role, content):
        self.messages.append({'role': role, 'content': content})

    def get_history(self):
        return self.messages
    
def query_with_messages(message_history, kind='coding', format='', temperature=0.5, top_p=0.8, top_k=20):
    # match kind:
        # case 'coding':
        #     model = 'qwen3-coder:30b'
        #     think = False
        # case 'simple':
        #     model = 'gpt-oss:20b'
        #     think = False
        # case 'reasoning':
        #     model = 'gpt-oss:20b'
        #     think = 'high'
    # model = 'Gemma4-26B-IQ4_NL:latest'
    model = 'qwen3-coder:30b'
    # model = 'Qwen3.6'
    think = False

    response = ol.chat(
        model=model,
        messages=message_history.get_history(),
        format=format,
        think=think,
        options = {
            "temperature": temperature,
            "top_p": top_p,        
            "top_k": top_k,          
            "num_ctx": 8192,      
            "repeat_penalty": 1.1, 
            "num_predict": 8192,  
            "mirostat": 0         
        }
    )

    print(response)
    return response['message']['content']
````

````python
class Transform(pydantic.BaseModel):
    model_config = pydantic.ConfigDict(arbitrary_types_allowed=True) # Add this line
    position: Optional[Vector2] = None
    rotation: Optional[float] = None
    scale: Optional[Vector2] = None
    zindex: Optional[int] = None

class SceneObject(pydantic.BaseModel):
    object_id: str = pydantic.Field(description="Unique identifier for the object, e.g., 'table_01'")
    properties: Dict[str, Any] = pydantic.Field(description="Physical attributes like color, size, or position")
    transform: Transform = pydantic.Field(description="Transformations to apply to the object, e.g., {'position': {'x': 100, 'y': 200}, 'rotation': 45.0, 'scale': {'x': 1.0, 'y': 1.0}}")

    def render(self):
        print(f"Rendering {self.object_id} with props: {self.properties}")

class Scene(pydantic.BaseModel):
    model_config = pydantic.ConfigDict(arbitrary_types_allowed=True)
    
    objects: List[Any] = []
    width: int = 1000
    height: int = 1000

    def add_object(self, scene_object):
        self.objects = [obj for obj in self.objects if getattr(obj, 'object_id', None) != getattr(scene_object, 'object_id', None)]
        self.objects.append(scene_object)

    def render(self):
        sorted_objects = sorted(self.objects, key=lambda obj: getattr(obj.transform, 'zindex', 0) if obj.transform else 0)

        for obj in sorted_objects:
            rl.rl_push_matrix()
            rl.rl_translatef(obj.transform.position.x, obj.transform.position.y, 0)
            rl.rl_rotatef(obj.transform.rotation, 0, 0, 1)
            rl.rl_scalef(obj.transform.scale.x, obj.transform.scale.y, 1)

            obj.render()

            rl.rl_pop_matrix()
````

````python
def sysprompt_instantiate_objects(chat_history: MessageHistory):
    return f"""
You are a helpful assistant that generates Python code to create and manipulate a 2D scene using custom primitives.
The canvas is 1000x1000 pixels, and the origin (0, 0) is at the top-left corner. The x-axis extends to the right and the y-axis extends downwards.

Each object instantiation has four main components:
1) class_name: A string that specifies the name of the SceneObject subclass to instantiate, e.g., 'Circle'. The class must have already been defined.
2) object_id: A unique identifier for the object, e.g., 'circle_01'.
3) transform: A dictionary of attributes that define the transform of the object (position, rotation, scale, zindex).
4) class properties: A dictionary of attributes specific to the class used (e.g., color, size).

Your objective is to analyze the user query and determine if it requires the instantiation of new objects to fulfill the request.
These are the currently defined classes: {dump_classes()}.

These are the currently instantiated objects: {dump_objects()}.

This is our chat history so far:
{json.dumps(chat_history.get_history(), indent=2)}

If so, generate the necessary object instantiations in the specified format.
Each object instantiation should include the class_name of the object to create, a unique object_id, the transform, and a set of properties with their values.
The properties must match the expected properties for the class being instantiated. For example, if you are instantiating a 'Circle' class that has properties 'radius' and 'color', you need to provide values for those properties in the instantiation.

You have a 'instantiate' function available that takes care of creating the object and adding it to the scene which you have
to use in the 'code' output to generate the instances and setup the dependencies between them based on the user request. For example, if the user request is "I want a blue square on top of a yellow rectangle", you might generate code that first instantiates a yellow rectangle and then instantiates a blue square with properties that position it on top of the rectangle.

The instantiate function has the following signature:
def instantiate_object(class_name: str, object_id: str, transform: Transform, properties: Dict[str, Any]):
And it can be used with the following syntax in the code output:
instantiate_object(class_name='Rectangle', object_id='rect_01', transform=Transform(position=Vector2(x=100, y=200), rotation=0.0, scale=Vector2(x=1.0, y=1.0), zindex=0), properties={{'width': 300, 'height': 150, 'color': (255, 255, 0)}})
According to the instances that you decided to create.

CRITICAL SPATIAL REASONING RULES:
Do NOT hardcode X and Y coordinates if an object's position depends on another object.
Instead, use Python variables in your 'code' output to calculate bounding boxes.

For example, to place 'blue_square' on top of 'skyscraper':
# Create the base object first
sky_width = 200
sky_height = 400
sky_x = 400
sky_y = 600
instantiate_object(class_name='Skyscraper', object_id='sky_01', transform=Transform(position=Vector2(x=sky_x, y=sky_y), rotation=0.0, scale=Vector2(x=1.0, y=1.0), zindex=0), properties={{'width': sky_width, 'height': sky_height, ...}})

# Calculate the position for the top object mathematically
square_size = 100
# Center X aligns with skyscraper Center X
square_x = sky_x 
# Y is calculated: Skyscraper Center Y - Half Skyscraper Height - Half Square Height
square_y = sky_y - (sky_height / 2) - (square_size / 2) 

CRITICAL SCHEMA RULES:
1. You MUST ONLY output properties that exactly match the provided class definition. Discard any properties the user requested if they do not exist in the class definition (e.g., do not output 'shape' if it's not in the dictionary).
2. You MUST output ALL properties required by the class. If the user didn't specify 'width', 'height', or 'radius', you must invent reasonable default float values.
3. Convert all color names to Tuple[int, int, int] (e.g., map 'red' to [255, 0, 0]).

instantiate_object(class_name='Rectangle', object_id='square_01', transform=Transform(position=Vector2(x=square_x, y=square_y), rotation=0.0, scale=Vector2(x=1.0, y=1.0), zindex=0), properties={{'width': square_size, 'height': square_size, ...}})

You have the following methods available to change the transform and properties of the objects after instantiation if needed:
{llm_instance_primitives_str}

You MUST only write the code to instantiate the objects and set up their properties and transforms. Do NOT write any code for rendering or defining classes, as those are handled separately. Your code should focus solely on creating the necessary objects with the correct properties and spatial relationships to fulfill the user request based on the defined classes and the current state of the scene.
All the class and library dependencies are already handled, so you can directly use the class names and the instantiate_object function without worrying about imports or definitions.
"""

def userprompt_instantiate_objects(user_query):
    return f"""User request: '{user_query}'
"""

def dump_objects():
    result = []
    for obj in scene.objects:
        result.append(str(obj.object_id) + ": " + obj.__class__.__name__  + ": " + str(obj.properties))
    return ',\n'.join(result)

def instantiate_object(class_name: str, object_id: str, transform: Transform, properties: Dict[str, Any]):
    instance_def = LLMResponse_ObjectCreationParameters(
        class_name=class_name,
        object_id=object_id,
        transform=transform,
        properties=properties
    )
    instantiate_object_structured(instance_def)

def instantiate_object_structured(instance_def: LLMResponse_ObjectCreationParameters):
    TargetClass = globals().get(instance_def.class_name)
    
    if not TargetClass:
        raise ValueError(f"Class {instance_def.class_name} has not been defined.")

    new_obj = TargetClass(
        object_id=instance_def.object_id, 
        transform=instance_def.transform,
        **instance_def.properties
    )
    
    scene.add_object(new_obj)
    print(f"Created {instance_def.class_name} with ID: {instance_def.object_id}")
````

````python
class System:
    def __init__(self, class_def_retries=3):
        self.interaction_history = MessageHistory()
        self.class_def_retries = class_def_retries


    def preprocessing(self, user_query) -> LLMResponse_PreprocessedQuery_List:
        messages = MessageHistory()
        messages.add_message('system', sysprompt_preprocess_request(self.interaction_history))
        messages.add_message('user', userprompt_preprocess_request(user_query))

        print ("--- Messages sent to LLM for Preprocessing ---")
        for msg in messages.get_history():
            print (f"{msg['role']}: {msg['content']}")

        response = query_with_messages(messages, kind='reasoning', temperature=1, format=LLMResponse_PreprocessedQuery_List.model_json_schema())
        print ("--- Preprocessing LLM Response ---")
        print (response)

        return LLMResponse_PreprocessedQuery_List.model_validate(json.loads(response))

    def class_identification_description(self, user_query: str) -> LLMResponse_ClassDescriptions_List:
        messages = MessageHistory()
        messages.add_message('system', sysprompt_class_identification_description())
        messages.add_message('user', userprompt_class_identification_description(user_query))

        print ("--- Messages sent to LLM for Class Identification Description ---")
        for msg in messages.get_history():
            print (f"{msg['role']}: {msg['content']}")

        response = query_with_messages(messages, kind='reasoning', temperature=0.5, format=LLMResponse_ClassDescriptions_List.model_json_schema())
        print ("--- Raw LLM Response for Class Identification Description ---")
        print (response)
        return LLMResponse_ClassDescriptions_List.model_validate(json.loads(response))

    def class_definition(self, user_query, messages: MessageHistory = None) -> Tuple[LLMResponse_ClassDefinitionParameters, MessageHistory]:
        if messages is None:
            messages = MessageHistory()
            messages.add_message('system', sysprompt_ask_for_new_classes())

        messages.add_message('user', userprompt_ask_for_new_classes(user_query))

        print ("--- Messages sent to LLM ---")
        for msg in messages.get_history():
            print (f"{msg['role']}: {msg['content']}")

        response = query_with_messages(messages, format=LLMResponse_ClassDefinitionParameters.model_json_schema(),\
                                        temperature=0.1, top_k=10, top_p=0.1)

        print ("--- Raw LLM Response ---")
        print (response)

        messages.add_message('assistant', response)

        return (LLMResponse_ClassDefinitionParameters.model_validate(json.loads(response)), messages)

    def class_change(self, user_query: str) -> LLMResponse_ClassDefinitionParameters:
        messages = MessageHistory()
        messages.add_message('system', sysprompt_modify_existing_class())
        messages.add_message('user', userprompt_modify_existing_class(user_query))

        print ("--- Messages sent to LLM for Class Modification ---")
        for msg in messages.get_history():
            print (f"{msg['role']}: {msg['content']}")

        response = query_with_messages(messages, format=LLMResponse_ClassDefinitionParameters.model_json_schema(),\
                                        temperature=0.1)

        print ("--- Raw LLM Response for Class Modification ---")
        print (response)

        return LLMResponse_ClassDefinitionParameters.model_validate(json.loads(response))

    def class_evaluation(self, description: str, class_def: LLMResponse_ClassDefinitionParameters) -> str:
        messages = MessageHistory()
        messages.add_message('system', sysprompt_class_evaluation())
        messages.add_message('user', userprompt_class_evaluation(description, class_def))

        print ("--- Messages sent to LLM for Class Evaluation ---")
        for msg in messages.get_history():
            print (f"{msg['role']}: {msg['content']}")

        response = query_with_messages(messages, kind='reasoning', 
                                        temperature=0.1)

        print ("--- Raw LLM Response for Class Evaluation ---")
        print (response)

        return response

    def object_instantiations(self, user_query) -> LLMResponse_ObjectCreationParameters_List:
        messages = MessageHistory()
        messages.add_message('system', sysprompt_instantiate_objects(self.interaction_history))
        messages.add_message('user', userprompt_instantiate_objects(user_query))

        print ("--- Messages sent to LLM ---")
        for msg in messages.get_history():
            print (f"{msg['role']}: {msg['content']}")

        response = query_with_messages(messages, temperature=0.1, format=LLMResponse_ObjectCreationParameters_List.model_json_schema())

        self.interaction_history.add_message('user', f"{user_query}. Add all missing properties to properties")
        self.interaction_history.add_message('assistant', response)

        print ("--- Raw LLM Response ---")
        print (response)

        return LLMResponse_ObjectCreationParameters_List.model_validate(json.loads(response))

    def handle_exception(self, code_snippet: str, error_message: str) -> str:
        messages = MessageHistory()
        messages.add_message('system', sysprompt_execution_error())
        messages.add_message('user', userprompt_execution_error(code_snippet, error_message))

        print ("--- Messages sent to LLM for Error Handling ---")
        for msg in messages.get_history():
            print (f"{msg['role']}: {msg['content']}")

        response = query_with_messages(messages, format='')

        print ("--- LLM Response for Error Handling ---")
        print (response)

        stripped_response = strip_code_markers(response)

        print ("--- Stripped LLM Response for Error Handling ---")
        print (stripped_response)

        return stripped_response

    def send_query(self, user_query):
        sentences = self.preprocessing(user_query)
        
        for sentence in sentences.steps:
            print (f"Processing sentence with intent '{sentence.intent}': '{sentence.description}'")
            match sentence.intent:
                case 'create_class':
                    print(f"Processing sentence with intent 'create_class': '{sentence.description}'")

                    classes = self.class_identification_description(sentence.description)

                    for llm_class in classes.descriptions:
                        class_description_str = llm_class.class_name + ": " + ', '.join(llm_class.sentences)

                        print(f"Combined class description for '{llm_class.class_name}': {class_description_str}")

                        class_definition_query = f"{class_description_str}. Also include {sentence.description}."
                        class_def_messages = None

                        for i in range(self.class_def_retries):
                            print (f"--- Iteration {i+1} for class definition of '{llm_class.class_name}' ---")

                            class_def, class_def_messages = self.class_definition(class_definition_query, class_def_messages)
                            evaluation = self.class_evaluation(class_description_str, class_def)
                            if evaluation == "OK":
                                print(f"Class definition for '{class_def.class_name}' passed evaluation.")
                                break
                            else:
                                print(f"Class definition for '{class_def.class_name}' failed evaluation: {evaluation}")
                                
                                class_definition_query = userprompt_ask_for_fixes(evaluation)

                        while True:
                            try:
                                define_class(class_def)

                                self.interaction_history.add_message('user', class_description_str)
                                self.interaction_history.add_message('assistant', class_def.model_dump_json())

                                break
                            except Exception as e:
                                print(f"Error occurred while defining class {class_def.class_name}: {e}")
                                response = self.handle_exception(code_snippet=class_def.render_function_body, error_message=str(e))
                                print(f"LLM response for error handling: {response}")

                                class_def.render_function_body = response
                case 'create_objects':
                    object_instantiations = self.object_instantiations(sentence.description)
                    while True:
                        try:
                            exec(object_instantiations.code, globals(), {'instantiate_object': instantiate_object})
                            break
                        except Exception as e:
                            print(f"Error occurred while generating object instantiations: {e}")
                            response = self.handle_exception(code_snippet=object_instantiations.code, error_message=str(e))
                            print(f"LLM response for error handling: {response}")

                            object_instantiations.code = response
                case 'change_class':
                    class_def = self.class_change(sentence.description)
                    while True:
                        try:
                            define_class(class_def, add_to_list=True)
                            
                            for obj in scene.objects:
                                if obj.__class__.__name__ == class_def.class_name:
                                    obj.__class__ = globals()[class_def.class_name]
                                    
                            break
                        except Exception as e:
                            print(f"Error occurred while modifying class {class_def.class_name}: {e}")
                            response = self.handle_exception(code_snippet=class_def.render_function_body, error_message=str(e))
                            print(f"LLM response for error handling: {response}")

                            class_def.render_function_body = response
                case 'change_objects':
                    object_instantiations = self.object_instantiations(sentence.description)
                    while True:
                        try:
                            exec(object_instantiations.code, globals(), {'instantiate_object': instantiate_object})
                            break
                        except Exception as e:
                            print(f"Error occurred while generating object instantiations: {e}")
                            response = self.handle_exception(code_snippet=object_instantiations.code, error_message=str(e))
                            print(f"LLM response for error handling: {response}")

                            object_instantiations.code = response
                case 'acknowledge':
                    print(f"Acknowledging sentence: '{sentence.description}'")
                    self.interaction_history.add_message('user', sentence.description)
                    self.interaction_history.add_message('assistant', "Acknowledged.")
````
